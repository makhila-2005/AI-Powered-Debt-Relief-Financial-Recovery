```python
import sqlite3
import google.generativeai as genai
```

```python
conn = sqlite3.connect("loans.db")

cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS loans(
amount INTEGER,
emi INTEGER,
income INTEGER,
overdue INTEGER
)
""")

conn.commit()
```

```python
def analyze_debt(amount,emi,income,overdue):

    emi_ratio = emi/income

    if emi_ratio > 0.5:
        stress="High Debt Stress"
    elif emi_ratio > 0.3:
        stress="Medium Debt Stress"
    else:
        stress="Low Debt Stress"


    settlement = amount*0.65


    return {
        "Debt Stress":stress,
        "EMI Ratio":round(emi_ratio,2),
        "Recommended Settlement":settlement
    }
```

```python
result = analyze_debt(
    500000,
    25000,
    40000,
    90
)

result
```

```python
genai.configure(
api_key="YOUR_API_KEY"
)


model = genai.GenerativeModel(
"gemini-1.5-flash"
)


response=model.generate_content(
"""
Generate a professional debt settlement negotiation letter
for a borrower who cannot pay full EMI.
"""
)


print(response.text)
```
